## Layoff Risk by Industry

```python
industry_risk = pd.crosstab(
    df["Industry"],
    df["Layoff_Risk"],
    normalize="index"
)*100

industry_risk
```

### Visualization

```python
industry_risk.plot(
    kind="bar",
    stacked=True,
    figsize=(12,6)
)

plt.show()
```

### AI Training Impact

```python
# Convert target into score:

risk_map = {
    "Low":1,
    "Medium":2,
    "High":3
}

df["Layoff_Risk_Score"] = df["Layoff_Risk"].map(risk_map)
```

```python
df.groupby("Layoff_Risk")[
    "AI_Training_Hours"
].mean()
```

### Experience vs Layoff Risk

```python
df["Experience_Group"] = pd.cut(
    df["Years_of_Experience"],
    bins=[0,3,7,12,50],
    labels=[
        "0-3 Years",
        "4-7 Years",
        "8-12 Years",
        "12+ Years"
    ]
)
```

```python
pd.crosstab(
    df["Experience_Group"],
    df["Layoff_Risk"],
    normalize="index"
)*100
```

### Correlation Analysis

```python
corr_cols = [
    'Age',
    'Years_of_Experience',
    'Routine_Task_Percentage',
    'Creativity_Requirement',
    'Human_Interaction_Level',
    'Number_of_AI_Tools_Used',
    'AI_Usage_Hours_Per_Week',
    'Tasks_Automated_Percentage',
    'AI_Training_Hours',
    'Layoff_Risk_Score'
]

corr = df[corr_cols].corr()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(corr,
            annot=True)

plt.show()
```
